```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("heights_outliers.csv")
df.head(3)
df.describe()
```

```python
#By setting KDE (Kernal Density Estimator) we will see that extra  curve..
sns.histplot(df.height,kde=True)

mean = round(df.height.mean(),2)
std = round(df.height.std(),2)

print('mean=',mean)
print('std=',std)

plt.axvline(mean, color='k', linestyle='dashed', linewidth=1)

# one standard deviation
plt.axvline(mean + std, color='r', linestyle='dashed', linewidth=1)
plt.axvline(mean - std, color='r', linestyle='dashed', linewidth=1)

# two standard deviation
plt.axvline(mean + 2*std, color='g', linestyle='dashed', linewidth=1)
plt.axvline(mean - 2*std, color='g', linestyle='dashed', linewidth=1)

# three standard deviation
plt.axvline(mean + 3*std, color='b', linestyle='dashed', linewidth=1)
plt.axvline(mean - 3*std, color='b', linestyle='dashed', linewidth=1)

#Add legends
plt.legend({'Mean:':mean,' std=':std ,'mean + std=': mean + std, '(mean + 2*std)=' :(mean + 2*std), ' (mean + 3*std)=' :(mean + 3*std)})


# Show the plot
plt.show()
```

```python
lower = mean - 3 * std
upper = mean + 3 * std
lower, upper
```

```python
print(" Outliers , All those records which are below lower or greater than upper.")
df[(df.height < lower) | (df.height > upper)]
```

```python

df_no_outliers = df[(df.height > lower) & (df.height < upper)]
print(' Total records that are not outliers')
df_no_outliers.shape
```
